# `prepare_workspace`
Diese Methode soll die Eingaben vom Nutzer verwenden, um den Workspace für den Agenten vorzubereiten.

```python
from pathlib import Path
import sys
import logging

logging.basicConfig(level=logging.INFO)

ROOT = Path.cwd().parent.parent
logging.info(f"Adding {ROOT} to Python path for imports in kernel.")
sys.path.insert(0, str(ROOT))  # ROOT, nicht SRC!

# Load config with changed env path
from bxagent.config import Config
config = Config.get_instance(env_path=ROOT / ".env")
```

```python
from bxagent.preparation.prepare_workspace import create_prepare_workspace_node

prepare_workspace = create_prepare_workspace_node()
```

```python
from bxagent.preparation.state import PreparationState

state = PreparationState(
    required_commands=[],
    workspace_path=ROOT / ".bx-agent-workspace" / "preparation",
    package_path="de.hof-university.bxagent",
    latest_validation_results=[]
)

new_state = prepare_workspace(state)
```
