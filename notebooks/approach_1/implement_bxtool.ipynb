# implement_bxtool
This function uses a Jinja Template to generate the adapter `BXToolForEMF` to make the generated transformation class available for the benchmarx testsuite.

```python
from pathlib import Path
import sys
import logging

logging.basicConfig(level=logging.INFO)

ROOT = Path.cwd().parent.parent
logging.info(f"Adding {ROOT} to Python path for imports in kernel.")
sys.path.insert(0, str(ROOT))  # ROOT, nicht SRC!

# Load config with changed env path
from mdagent.config import Config
config = Config.get_instance(env_path=ROOT / ".env")
```

At first some prerequisities have to be made: init the LLM and set a workspace path.

```python
from mdagent.models import build_base_model
from pathlib import Path

coding_model = build_base_model()
workspace = ROOT / ".mdagent_workspace" 
```

After that, the node can be created.

```python
from mdagent.tools.coding import create_implement_bx_tool_node

implement_bx_tool = create_implement_bx_tool_node(llm=coding_model, workspace=workspace)
```

Now, we can test if the agent is able to implement the template accordingly.

```python
from mdagent.tools.coding import ImplementationState
from mdagent.tools.transformation import (
    TransformationPlan,
    FileTransformationPlanParser,
)

tp_md_path = ROOT / ".mdagent-workspace" / "TRANSFORMATION.md"
bxtool_file = ROOT / ".mdagent-workspace" / "bxTool.java"

tp = TransformationPlan.parse(
    parser=FileTransformationPlanParser(file_path=tp_md_path),
    template_path=ROOT / "templates"
)

state = ImplementationState(
    transformation_plan=tp,
    task_specification="Implement the bx tool as specified in the transformation plan. Assume that the transformation plan is complete and correct. Focus on implementing the bx tool according to the specifications in the transformation plan.",
    written_java_files=[],
    bxtool_file=bxtool_file,
    transformation_implementation="",
    latest_evaluation_results=[],
    implementation_iteration=0
)
```

```python
implement_bx_tool(state=state)
```
