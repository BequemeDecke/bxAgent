# implement_transformation
This node uses a deep agent to implement the transformation class according to the transformation plan (TRANSFORMATION.md)

```python
from pathlib import Path
import sys
import logging

logging.basicConfig(level=logging.INFO)

ROOT = Path.cwd().parent.parent
logging.info(f"Adding {ROOT} to Python path for imports in kernel.")
sys.path.insert(0, str(ROOT))  # ROOT, nicht SRC!

# Load config with changed env path
from bxagent.config import Config
config = Config.get_instance(env_path=ROOT / ".env")
```

At first some prerequisities have to be made: init the Coding Deep Agent and the transformation plan

```python
from bxagent.models import build_base_model
from bxagent.tools.coding import CodingAgentState
from bxagent.agents.coding.agent import build_coding_deep_agent
from bxagent.tools.transformation import (
    TransformationPlan,
    FileTransformationPlanParser,
)
from pathlib import Path

tp_md_path = ROOT / ".bx-agent-workspace" / "TRANSFORMATION.md"
bxtool_file = ROOT / ".bx-agent-workspace" / "bxTool.java"

tp = TransformationPlan.parse(
    parser=FileTransformationPlanParser(file_path=tp_md_path),
    template_path=ROOT / "templates"
)

coding_agent = build_coding_deep_agent()
```

After that, the node can be created.

```python
from bxagent.tools.coding import create_implement_transformation_node

implement_transformation = create_implement_transformation_node(coding_agent=coding_agent, optional_plan_factory=lambda: tp)
```

Now, we can test if the agent is able to implement the template accordingly.

```python
from bxagent.tools.coding import CodingAgentState

state = CodingAgentState(
    transformation_plan=tp,
    task_specification="Implement the transformation class as specified in the transformation plan.",
    written_java_files=[],
    bxtool_file=bxtool_file,
    transformation_implementation="",
    latest_validation_results=[],
    implementation_iteration=0
)
```

```python
response = implement_transformation(state)
```
